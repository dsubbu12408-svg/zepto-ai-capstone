```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("titanic")

print("Dataset loaded successfully")
print("Shape:", df.shape)

df.head()
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
print("Data Types:")
print(df.dtypes)

print("\nSummary Statistics:")
display(df.describe())
```

```python
print("Unique Values:")
print(df.nunique())
```

```python
import matplotlib.pyplot as plt

numeric_df = df.select_dtypes(include="number")

plt.figure(figsize=(10, 6))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
print("Survival Rate by Sex:")
display(df.groupby("sex")["survived"].mean())
```

```python
print("Survival Rate by Passenger Class:")
display(df.groupby("pclass")["survived"].mean())
```

```python
print("Average Age by Survival:")
display(df.groupby("survived")["age"].mean())
```

```python
print("Survival Rate by Sex and Passenger Class:")
display(
    df.groupby(["sex", "pclass"])["survived"].mean()
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(df["age"].dropna(), bins=20)
plt.title("Distribution of Passenger Age")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="pclass", y="fare")
plt.title("Fare Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="sex", y="survived")
plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
corr = df[["age", "fare", "pclass", "survived"]].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

## EDA Findings and Interpretations

### 1. Age Distribution
The age distribution is concentrated mainly around young and middle-aged passengers, with fewer passengers at older ages. The histogram also shows some very young passengers and a smaller number of elderly passengers.

### 2. Fare Distribution by Passenger Class
The box plot shows that Passenger Class 1 had substantially higher fares than Classes 2 and 3. Class 1 also contains several high-fare outliers.

### 3. Survival Rate by Sex
The survival comparison shows a much higher survival rate among female passengers than male passengers in the Titanic dataset.

### 4. Correlation Heatmap
The heatmap shows a moderate negative correlation between passenger class and fare (-0.55), and between passenger class and survival (-0.34). Fare has a positive correlation with survival (0.26), while age has very little correlation with survival (-0.08).

## Conclusion

The EDA shows that passenger class, fare, and sex are important factors associated with survival in the Titanic dataset. Passenger class and fare are moderately related to survival, while age has very little correlation with survival.
